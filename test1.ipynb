```kotlin
//점수가 90점 이상이면 A, 80점 이상이면 B, 아니면 C로 출력하는 if 식을 작성하세요.
fun checkGrade(x:Int){
    if(x >= 90){
        println("A")
    } else if(x >= 80){
        println("B")
    }else{
        println("C")
    }
}

checkGrade(50);
checkGrade(80);
checkGrade(92);
```
